```python


pd.set_option("display.max_colwidth", 80)

PARENT_ID = "EAS_001"   # Giza Pyramids and The Sphinx
RADIUS_KM = 1.5         # distance around the pyramids plateau

parent = data[data["place_id"] == PARENT_ID].iloc[0]
plat, plon = float(parent["latitude"]), float(parent["longitude"])

def haversine_km(lat1, lon1, lat2, lon2):
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])
    a = np.sin((lat2 - lat1) / 2) ** 2 + np.cos(lat1) * np.cos(lat2) * np.sin((lon2 - lon1) / 2) ** 2
    return 6371 * 2 * np.arcsin(np.sqrt(a))

tmp = data.copy()
tmp["dist_km"] = haversine_km(
    plat, plon,
    pd.to_numeric(tmp["latitude"], errors="coerce"),
    pd.to_numeric(tmp["longitude"], errors="coerce"),
)

keywords = "pyramid|sphinx|khufu|cheops|khafre|chephren|menkaure|mycerinus|valley temple|queens"
by_name = tmp["place_name"].str.contains(keywords, case=False, na=False)
by_dist = (tmp["dist_km"] <= RADIUS_KM) & (tmp["city"] == "Giza")

candidates = tmp[(by_name | by_dist) & (tmp["place_id"] != PARENT_ID)]
display(candidates[["place_id", "place_name", "city", "dist_km"]].sort_values("dist_km"))
```

```python
TAG_KEYWORDS = {
    "ancient":   ["ancient", "archaeological", "pharaonic", "necropolis", "pyramid", "nubian"],
    "temple":    ["temple", "sanctuary", "cult"],
    "tomb":      ["tomb", "mortuary", "valley of", "funerary"],
    "museum":    ["museum"],
    "unesco":    ["unesco"],
    "religious": ["mosque", "church", "monastery", "synagogue", "religious"],
    "nature":    ["nature", "park", "oasis", "desert", "reserve", "mountain"],
    "beach":     ["beach", "island", "sea", "reef"],
    "islamic":   ["islamic", "citadel", "mosque", "sabil", "kuttab", "sultan", "bab zuwayla"],
    "market":    ["bazaar", "market", "souk"],
    "monument":  ["monument", "statue", "obelisk", "colossi", "sphinx"],
    "heritage":  ["historic", "heritage", "landmark", "house", "mansion", "old city", "district", "quarter"],
    "palace":    ["palace", "castle", "fortress"],
}

def to_tags(text):
    t = str(text).lower()
    tags = [tag for tag, kws in TAG_KEYWORDS.items()
            if any(re.search(rf"\b{re.escape(k)}", t) for k in kws)]
    return tags or ["other"]

# مفيش عمود tag_text: بنحسب التاجز مباشرة من الاسم + الفئة
data["tags"] = (data["place_name"].astype(str) + " " + data["category"].astype(str)).apply(to_tags)
data["is_panorama"] = data["place_name"].str.contains("panorama", case=False).astype(int)
```
